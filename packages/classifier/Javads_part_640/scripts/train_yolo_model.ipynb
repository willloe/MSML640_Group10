```python
from ultralytics import YOLO

model = YOLO("yolov8s-obb.pt")

results = model.train(
    data='/content/logos-images/dataset_diagrams_ai2d/dataset.yaml',
    epochs=80,
    imgsz=1024,
    batch=-1,
    workers=4,
    device=0,                # use GPU
    patience=20,             # early stopping
    optimizer="auto",
    cos_lr=True,
    project="slide_components_obb_runs",
    name="yolov8s_obb_1024_b32",
    save_period=-1,
)
```
